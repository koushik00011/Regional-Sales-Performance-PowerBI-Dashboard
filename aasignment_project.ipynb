```python
import pandas as pd
```

```python
df2015=pd.read_csv(r'C:\my_da_projects\assignment_1st_round\order_2015.csv')
df2016=pd.read_csv(r'C:\my_da_projects\assignment_1st_round\order_2016.csv')
df2017=pd.read_csv(r'C:\my_da_projects\assignment_1st_round\order_2017.csv')
df2018=pd.read_csv(r'C:\my_da_projects\assignment_1st_round\order_2018.csv')
```

```python
orders=orders.rename(columns={'Order ID':'order_id','Order Date':'order_date','Ship Date':'ship_date','Ship Mode':'ship_mode','Customer ID':'customer_id',
                     'Segment':'segment','City':'city','State':'state','Country':'country','Region':'region','Product ID':'product_id','Category':'category',
                     'Sub-Category':'sub_category','Product Name':'product_name','Sales':'sales','Quantity':'quantity','Discount':'discount','Profit':'profit',
                             'Customer Name':'customer_name'})

```

```python
orders.head()
```

```python
orders['order_date'] = pd.to_datetime(
    orders['order_date'],
    unit='D',
    origin='1899-12-30'
)

orders['ship_date'] = pd.to_datetime(
    orders['ship_date'],
    unit='D',
    origin='1899-12-30'
)
```

```python
map_data={'one':1,'two':2,'three':3,'four':4,'five':5,'six':6,'seven':7,'eight':8,'nine':9,'ten':10}
orders['quantity']=orders['quantity'].replace(map_data)
```

```python
orders['quantity']=pd.to_numeric(orders['quantity'])
```
